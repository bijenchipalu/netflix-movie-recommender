<h3>Vectorization</h3>

<h3>Stem Function</h3>

```python
# stem before vectorizationothere wise wrong
import nltk 
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
```

```python
def stem(text):
    x=[]
    for i in text.split():
        x.append(ps.stem(i))
    return " ".join(x) 
```

```python
movie['tag'] = movie['tag'].apply(stem)
```

```python
movie['tag'][0]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 5000,stop_words = 'english')
```

```python
vector = cv.fit_transform(movie['tag']).toarray()
```

```python
vector
```

```python
cv.get_feature_names_out()
```

<h3>Finding Distance between movie</h3>

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vector)
```

```python
similarity[1]
```

```python
similarity.shape
```

```python
sorted(list(enumerate(similarity[0])),reverse=True, key=lambda x:x[1])[1:6]
```

```python
def recommend(movie_name):
    index = movie[movie['title']==movie_name].index[0]
    distances = similarity[index]
    flim_list = sorted(list(enumerate(distances)),reverse = True, key = lambda x: x[1])[1:6]
    
    for i in flim_list:
        print(movie.iloc[i[0]].title)
```

```python
recommend('The Matrix Reloaded')
```
